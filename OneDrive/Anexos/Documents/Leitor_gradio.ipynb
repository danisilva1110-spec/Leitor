# Leitor de PDF com áudio no Colab

Este notebook usa o upload de arquivos do Colab para enviar um PDF e gerar um áudio com a leitura do texto.

```python
# Instale as dependências necessárias
!pip -q install pypdf gTTS
```

```python
from pathlib import Path

from gtts import gTTS
from pypdf import PdfReader
try:
    from google.colab import files
except ImportError as exc:
    raise ImportError("Execute este notebook no Google Colab para usar o upload de arquivos.") from exc


def extrair_texto_pdf(caminho_pdf: Path) -> str:
    leitor = PdfReader(str(caminho_pdf))
    return "".join(pagina.extract_text() or "" for pagina in leitor.pages).strip()


def gerar_audio(texto: str, idioma: str, destino: Path) -> Path:
    if not texto:
        raise ValueError("Não foi possível extrair texto desse PDF.")
    tts = gTTS(text=texto, lang=idioma)
    tts.save(str(destino))
    return destino


print("Pronto! Execute a próxima célula para enviar o PDF.")
```

```python
# Envie o PDF pelo upload do Colab
uploaded = files.upload()
pdf_path = Path(next(iter(uploaded.keys())))

idioma = "pt"  # altere para 'en' ou 'es' se quiser
texto = extrair_texto_pdf(pdf_path)
audio_path = gerar_audio(texto, idioma, Path("leitura.mp3"))

print(f"Áudio salvo em: {audio_path}")
files.download(str(audio_path))
```
